# Exploratory Data Analysis of Student Performance Using Python

**Internship Task:** Task 3 – Exploratory Data Analysis (EDA)

**Objective:** To analyze student performance data using statistical summaries and visualizations, identify patterns and correlations, and understand factors associated with students' marks.

**Tools Used:** Python, Pandas, NumPy, Matplotlib, Seaborn, and Google Colab.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

data = {
    "Student_ID": range(1, 21),
    "Study_Hours": [
        1, 2, 3, 4, 5, 2, 6, 7, 3, 8,
        4, 5, 9, 6, 7, 2, 8, 10, 5, 9
    ],
    "Attendance": [
        55, 60, 65, 70, 75, 62, 80, 85, 68, 90,
        72, 78, 95, 82, 88, 58, 92, 98, 76, 94
    ],
    "Assignment_Score": [
        40, 45, 50, 55, 60, 48, 70, 75, 52, 82,
        58, 65, 90, 72, 78, 42, 85, 95, 63, 88
    ],
    "Exam_Marks": [
        35, 40, 45, 50, 58, 42, 68, 74, 48, 85,
        55, 62, 92, 70, 80, 38, 84, 96, 60, 90
    ]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
display(df.head())
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst five records:")
display(df.head())

print("\nLast five records:")
display(df.tail())
```

```python

print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Exam_Marks",
    s=100
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Daily Study Hours")
plt.ylabel("Exam Marks")
plt.tight_layout()
plt.show()
```

```python

print("Average Study Hours:",
      round(df["Study_Hours"].mean(), 2))

print("Average Attendance:",
      round(df["Attendance"].mean(), 2), "%")

print("Average Assignment Score:",
      round(df["Assignment_Score"].mean(), 2))

print("Average Exam Marks:",
      round(df["Exam_Marks"].mean(), 2))

print("Highest Exam Marks:",
      df["Exam_Marks"].max())

print("Lowest Exam Marks:",
      df["Exam_Marks"].min())
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Exam_Marks",
    bins=8,
    kde=True
)

plt.title("Distribution of Student Exam Marks")
plt.xlabel("Exam Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Exam_Marks",
    s=100
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Daily Study Hours")
plt.ylabel("Exam Marks")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam_Marks",
    s=100
)

plt.title("Attendance vs Exam Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Marks")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Assignment_Score",
    y="Exam_Marks",
    s=100
)

plt.title("Assignment Score vs Exam Marks")
plt.xlabel("Assignment Score")
plt.ylabel("Exam Marks")
plt.tight_layout()
plt.show()
```

```python

numeric_df = df[
    ["Study_Hours", "Attendance",
     "Assignment_Score", "Exam_Marks"]
]

correlation = numeric_df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Student Performance")
plt.tight_layout()
plt.show()
```

```python

exam_correlation = correlation["Exam_Marks"].drop(
    "Exam_Marks"
).sort_values(ascending=False)

print("Correlation with Exam Marks:")
display(exam_correlation)

print("\nMost positively correlated factor:")
print(exam_correlation.idxmax())

print("\nCorrelation value:")
print(round(exam_correlation.max(), 3))
```

```python


def categorize_performance(marks):
    if marks < 40:
        return "Low"
    elif marks < 70:
        return "Medium"
    else:
        return "High"


df["Performance_Level"] = df["Exam_Marks"].apply(
    categorize_performance
)

print("Student Performance Categories:")
display(df[["Student_ID", "Exam_Marks", "Performance_Level"]])

print("\nNumber of students in each category:")
display(df["Performance_Level"].value_counts())

# Bar chart
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Performance_Level",
    order=["Low", "Medium", "High"]
)

plt.title("Number of Students by Performance Level")
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

print("========== KEY FINDINGS ==========")

print("Total students:", len(df))

print("Average exam marks:",
      round(df["Exam_Marks"].mean(), 2))

print("Highest exam marks:",
      df["Exam_Marks"].max())

print("Lowest exam marks:",
      df["Exam_Marks"].min())

print("Average study hours:",
      round(df["Study_Hours"].mean(), 2))

print("Average attendance:",
      round(df["Attendance"].mean(), 2), "%")

print("\nStudents in each performance category:")
print(df["Performance_Level"].value_counts())

print("\nCorrelation with exam marks:")
print(correlation["Exam_Marks"].drop("Exam_Marks")
      .sort_values(ascending=False).round(3))
```
